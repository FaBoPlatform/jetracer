# 走行までの流れ

- 01_find_pwm PWMの値の設定をおこないます。
- 11_record_camera 1カメラで走行した履歴を保存します。
- 13_annotation 11で作成した走行履歴からアノテーションを実施します。
- 16_convert 13で保存した学習済みモデルを、TensorRT(軽量化)フォーマットに変換します。
- 17_run 16で変換したTensorRTモデルで走行します。

入門者向けの簡易チュートリアルは<br>
[https://faboplatform.github.io/JetracerDocs/quick/01.intro/](https://faboplatform.github.io/JetracerDocs/quick/01.intro/)<br>
に公開しています。

# 2カメラバージョン

今回から、2カメラを搭載したバージョンのサンプルの配布も開始しています。
1カメラ目(cam0)は走行用に使用し、2カメラ目(cam1)は環境情報用に使用する事が可能となります。

- 20_recored_dual_cam 2カメラでの走行した履歴を保存します。
- 30_detect cam1の撮影履歴をもとに環境情報のクラス分類を実施します。
- 31_convert_detect 30で保存した学習済みモデルとTensorRT(軽量化)フォーマットに変換します。
- 13_annotation 11で作成した走行履歴からアノテーションを実施します。
- 16_convert 13で保存した学習済みモデルを、TensorRT(軽量化)フォーマットに変換します。
- 40_run_dual_camera 13と16で作成した2つの走行モデルと、30と31で作成した1つの環境情報モデルから自動走行します。
- 41_dual_movie 40での実機での走行以外に、41で動画ベースで評価も可能です。


# 注意事項

WiFi環境によって、13_annotation, 30_detect を実行した際のデータセットの再生速度が遅くなる場合があります。sleep_timeをうまく調整してください。USBでPCとダイレクトに接続すれば、遅延はほとんど発生しなくなります。
